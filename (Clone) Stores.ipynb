# Filter data in python




E.g filter people older then 30 :
df[["age"]>30]

filter by an exact value
df[df["city"]=="JHB"]


df[(df["age"]>25)&(df["salary"]>300000)]
df[df["age"]==="JHB")|(df["age"]>=30)]

Filter using a list of values
df[df["city"].isin(["JHB","CPT"])]

Using the not filter(~) 
df[~df["city"].isin(["JHB","CPT"])]
This will give your city names that are not JHB or CPT

Filter data within a specific range
df[df("age").between(0,30)]

Filter rows where there is a missing value
df[df["salary"].isna()]

df[df["salary"].notna()]
filters rows where salary is available
